**# PART A: TABULAR MEDICAL DATA**

```python

# PART A: TABULAR MEDICAL DATA

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import zipfile
from scipy import stats
from scipy.stats import chi2_contingency, f_oneway

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif

# 2. LOAD DATASET

zip_path="/content/archive (8).zip"
unzipped_path="/content/diabetes"
zip_ref=zipfile.ZipFile(zip_path,"r")
zip_ref.extractall(unzipped_path)

df = pd.read_csv("diabetes/diabetes.csv")

print("Dataset loaded successfully!")
print("Shape of dataset:", df.shape)

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 3. BASIC DATA INSPECTION
# ============================================================

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe())

# 4. CHECK DUPLICATE RECORDS

duplicate_count = df.duplicated().sum()

print("\nNumber of duplicate rows:", duplicate_count)

if duplicate_count > 0:
    df = df.drop_duplicates()
    print("Duplicates removed.")
else:
    print("No duplicate records found.")



print("\nMissing values before cleaning:")
print(df.isnull().sum())


# In this dataset, zero can represent a missing measurement
# for several medical variables.

zero_as_missing = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

print("\nClinically suspicious zero values:")

for column in zero_as_missing:
    count = (df[column] == 0).sum()
    print(f"{column}: {count}")

# 7. REPLACE INVALID ZEROS WITH NaN

df_clean = df.copy()

for column in zero_as_missing:
    df_clean[column] = df_clean[column].replace(0, np.nan)

print("\nMissing values after converting invalid zeros:")
print(df_clean.isnull().sum())


# 8. HANDLE MISSING VALUES USING MEDIAN

for column in zero_as_missing:
    median_value = df_clean[column].median()
    df_clean[column] = df_clean[column].fillna(median_value)

print("\nMissing values after median imputation:")
print(df_clean.isnull().sum())


# 9. CHECK FINAL DATASET AFTER CLEANING

print("\nCleaned dataset shape:", df_clean.shape)

print("\nCleaned dataset:")
display(df_clean.head())

print("\nStatistical summary after cleaning:")
display(df_clean.describe())

# 10. CHECK TARGET / CLASS DISTRIBUTION

print("\nDiabetes outcome distribution:")
print(df_clean["Outcome"].value_counts())

print("\nPercentage distribution:")
print(df_clean["Outcome"].value_counts(normalize=True) * 100)


# Plot class distribution

plt.figure(figsize=(6, 4))

sns.countplot(
    x="Outcome",
    data=df_clean
)

plt.title("Diabetes Outcome Distribution")
plt.xlabel("Outcome (0 = No Diabetes, 1 = Diabetes)")
plt.ylabel("Number of Patients")
plt.show()

# 11. EDA — HISTOGRAMS

df_clean.hist(
    figsize=(14, 12),
    bins=20
)

plt.suptitle(
    "Distribution of Medical Features",
    fontsize=16
)

plt.tight_layout()
plt.show()

# 12. EDA — BOX PLOTS

numeric_features = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age"
]

plt.figure(figsize=(14, 7))

sns.boxplot(
    data=df_clean[numeric_features]
)

plt.title("Boxplots of Medical Features")
plt.xticks(rotation=45)
plt.ylabel("Value")
plt.show()


# 13. OUTLIER DETECTION USING IQR

print("\nOutlier counts using IQR method:")

for column in numeric_features:

    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[column] < lower_bound) |
        (df_clean[column] > upper_bound)
    ]

    print(f"{column}: {len(outliers)} outliers")


# 14. CORRELATION ANALYSIS

correlation_matrix = df_clean.corr(numeric_only=True)

plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Matrix")
plt.show()

# 15. GLUCOSE VS DIABETES

plt.figure(figsize=(7, 5))

sns.boxplot(
    x="Outcome",
    y="Glucose",
    data=df_clean
)

plt.title("Glucose Levels by Diabetes Outcome")
plt.xlabel("Outcome (0 = No Diabetes, 1 = Diabetes)")
plt.ylabel("Glucose")

plt.show()

# 16. BMI VS DIABETES

plt.figure(figsize=(7, 5))

sns.boxplot(
    x="Outcome",
    y="BMI",
    data=df_clean
)

plt.title("BMI by Diabetes Outcome")
plt.xlabel("Outcome")
plt.ylabel("BMI")

plt.show()


# 17. FEATURE ENGINEERING — BMI CATEGORY

df_clean["BMI_Category"] = pd.cut(
    df_clean["BMI"],
    bins=[0, 18.5, 25, 30, np.inf],
    labels=[
        "Underweight",
        "Normal",
        "Overweight",
        "Obese"
    ]
)

print("\nBMI categories:")
display(
    df_clean[["BMI", "BMI_Category"]].head(10)
)

# 18. FEATURE ENGINEERING — AGE GROUP

df_clean["Age_Group"] = pd.cut(
    df_clean["Age"],
    bins=[0, 30, 50, np.inf],
    labels=[
        "Young",
        "Middle-aged",
        "Older"
    ]
)

print("\nAge groups:")
display(
    df_clean[["Age", "Age_Group"]].head(10)
)


# ============================================================
# 19. HYPOTHESIS TEST
# Independent two-sample t-test
#
# Question:
# Is mean glucose level different between diabetic
# and non-diabetic patients?
# ============================================================

diabetic = df_clean[
    df_clean["Outcome"] == 1
]["Glucose"]

non_diabetic = df_clean[
    df_clean["Outcome"] == 0
]["Glucose"]


t_statistic, p_value = stats.ttest_ind(
    diabetic,
    non_diabetic,
    equal_var=False
)

print("\n========== HYPOTHESIS TEST ==========")

print("Mean glucose - Diabetic:",
      round(diabetic.mean(), 2))

print("Mean glucose - Non-diabetic:",
      round(non_diabetic.mean(), 2))

print("T-statistic:",
      round(t_statistic, 4))

print("P-value:",
      p_value)


alpha = 0.05

if p_value < alpha:
    print("\nConclusion:")
    print("Reject H0.")
    print(
        "There is a statistically significant difference "
        "in mean glucose levels between diabetic and "
        "non-diabetic patients."
    )
else:
    print("\nConclusion:")
    print("Fail to reject H0.")
    print(
        "There is no statistically significant difference "
        "in mean glucose levels between the two groups."
    )


# ============================================================
# 20. CHI-SQUARE TEST
#
# Question:
# Is BMI category associated with diabetes outcome?
# ============================================================

contingency_table = pd.crosstab(
    df_clean["BMI_Category"],
    df_clean["Outcome"]
)

print("\n========== CHI-SQUARE TEST ==========")

print("\nContingency table:")
display(contingency_table)


chi2_stat, chi2_p, degrees_of_freedom, expected = (
    chi2_contingency(contingency_table)
)

print("Chi-square statistic:",
      round(chi2_stat, 4))

print("Degrees of freedom:",
      degrees_of_freedom)

print("P-value:",
      chi2_p)


if chi2_p < alpha:
    print("\nConclusion:")
    print("Reject H0.")
    print(
        "BMI category and diabetes outcome "
        "are statistically associated."
    )
else:
    print("\nConclusion:")
    print("Fail to reject H0.")
    print(
        "No statistically significant association "
        "was found between BMI category and diabetes outcome."
    )


# ============================================================
# 21. ANOVA
#
# Question:
# Are mean glucose levels different among age groups?
# ============================================================

young = df_clean[
    df_clean["Age_Group"] == "Young"
]["Glucose"]

middle_aged = df_clean[
    df_clean["Age_Group"] == "Middle-aged"
]["Glucose"]

older = df_clean[
    df_clean["Age_Group"] == "Older"
]["Glucose"]


f_statistic, anova_p = f_oneway(
    young,
    middle_aged,
    older
)

print("\n========== ANOVA TEST ==========")

print("Young group mean:",
      round(young.mean(), 2))

print("Middle-aged group mean:",
      round(middle_aged.mean(), 2))

print("Older group mean:",
      round(older.mean(), 2))

print("\nF-statistic:",
      round(f_statistic, 4))

print("P-value:",
      anova_p)


if anova_p < alpha:
    print("\nConclusion:")
    print("Reject H0.")
    print(
        "At least one age group has a significantly "
        "different mean glucose level."
    )
else:
    print("\nConclusion:")
    print("Fail to reject H0.")
    print(
        "There is no statistically significant difference "
        "in mean glucose levels among the age groups."
    )


# ============================================================
# 22. FEATURE SELECTION
# ============================================================

# Remove engineered categorical columns for this analysis

X = df_clean[
    [
        "Pregnancies",
        "Glucose",
        "BloodPressure",
        "SkinThickness",
        "Insulin",
        "BMI",
        "DiabetesPedigreeFunction",
        "Age"
    ]
]

y = df_clean["Outcome"]


# Select top 5 features

selector = SelectKBest(
    score_func=f_classif,
    k=5
)

X_selected = selector.fit_transform(
    X,
    y
)

selected_features = X.columns[
    selector.get_support()
]

print("\n========== FEATURE SELECTION ==========")

print("Selected features:")
print(selected_features.tolist())


# Display feature scores

feature_scores = pd.DataFrame({
    "Feature": X.columns,
    "ANOVA_F_Score": selector.scores_
})

feature_scores = feature_scores.sort_values(
    by="ANOVA_F_Score",
    ascending=False
)

print("\nFeature scores:")
display(feature_scores)


# ============================================================
# 23. FEATURE SELECTION VISUALIZATION
# ============================================================

plt.figure(figsize=(10, 5))

sns.barplot(
    x="ANOVA_F_Score",
    y="Feature",
    data=feature_scores
)

plt.title("Feature Importance Based on ANOVA F-Score")
plt.xlabel("F-Score")
plt.ylabel("Feature")

plt.show()


# ============================================================
# 24. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\n========== TRAIN TEST SPLIT ==========")

print("Training features:",
      X_train.shape)

print("Testing features:",
      X_test.shape)

print("Training target:",
      y_train.shape)

print("Testing target:",
      y_test.shape)


# ============================================================
# 25. FEATURE SCALING
# ============================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)


print("\n========== FEATURE SCALING ==========")

print("Training data after scaling:")
display(
    pd.DataFrame(
        X_train_scaled,
        columns=X_train.columns
    ).head()
)


# ============================================================
# 26. CHECK SCALED FEATURE STATISTICS
# ============================================================

scaled_train_df = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns
)

print("\nMean after scaling:")
print(scaled_train_df.mean().round(2))

print("\nStandard deviation after scaling:")
print(scaled_train_df.std().round(2))


# ============================================================
# 27. FINAL MODEL-READY DATA
# ============================================================

print("\n========== FINAL MODEL-READY DATA ==========")

print("X_train shape:",
      X_train_scaled.shape)

print("X_test shape:",
      X_test_scaled.shape)

print("y_train shape:",
      y_train.shape)

print("y_test shape:",
      y_test.shape)

print("\nNo missing values in training data:",
      np.isnan(X_train_scaled).sum() == 0)

print("No missing values in testing data:",
      np.isnan(X_test_scaled).sum() == 0)


# ============================================================
# 28. FINAL SUMMARY
# ============================================================

print("\n==============================================")
print("          TABULAR DATA ANALYSIS COMPLETE")
print("==============================================")

print("\nOriginal dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)

print("\nProcessing performed:")
print("1. Data acquisition")
print("2. Data inspection")
print("3. Duplicate checking")
print("4. Missing-value detection")
print("5. Clinical invalid-value handling")
print("6. Median imputation")
print("7. Class imbalance analysis")
print("8. Exploratory data analysis")
print("9. Outlier analysis")
print("10. Correlation analysis")
print("11. Feature engineering")
print("12. Hypothesis testing")
print("13. Chi-square test")
print("14. ANOVA")
print("15. Feature selection")
print("16. Feature scaling")
print("17. Train-test splitting")
print("18. Model-ready data preparation")
```

**#Part-B: Text Data**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import zipfile
import os
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from collections import Counter

zip_path="/content/archive (9).zip"
unzipped_path="/content/mtsamples"
zip_ref=zipfile.ZipFile(zip_path,"r")
zip_ref.extractall(unzipped_path)

df = pd.read_csv("mtsamples/mtsamples.csv")

print("Dataset shape:", df.shape)
display(df.head())

print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()

df = df.dropna(subset=["transcription"])

df["transcription"] = df["transcription"].astype(str)

print("\nShape after cleaning:", df.shape)

print("\nMedical specialty distribution:")
print(df["medical_specialty"].value_counts())

plt.figure(figsize=(12, 6))
sns.countplot(
    y="medical_specialty",
    data=df,
    order=df["medical_specialty"].value_counts().head(15).index
)
plt.title("Top Medical Specialties")
plt.xlabel("Number of Records")
plt.ylabel("Medical Specialty")
plt.show()

df["text_length"] = df["transcription"].apply(len)

df["word_count"] = df["transcription"].apply(
    lambda x: len(x.split())
)

print("\nText length statistics:")
display(df[["text_length", "word_count"]].describe())

plt.figure(figsize=(8, 5))
sns.histplot(df["word_count"], bins=40, kde=True)
plt.title("Distribution of Document Word Count")
plt.xlabel("Number of Words")
plt.ylabel("Number of Documents")
plt.show()

def remove_phi(text):
    text = re.sub(
        r"\b[\w\.-]+@[\w\.-]+\.\w+\b",
        "[EMAIL]",
        text
    )
    text = re.sub(
        r"\b\d{3}[-.\s]?\d{3}[-.\s]?\d{4}\b",
        "[PHONE]",
        text
    )
    text = re.sub(
        r"\b\d{1,2}[/-]\d{1,2}[/-]\d{2,4}\b",
        "[DATE]",
        text
    )
    text = re.sub(
        r"\b\d{5}(?:-\d{4})?\b",
        "[ZIP]",
        text
    )
    return text

df["clean_text"] = df["transcription"].apply(remove_phi)

def clean_text(text):
    text = text.lower()
    text = re.sub(r"\s+", " ", text)
    text = re.sub(r"[^a-zA-Z0-9\s\[\]]", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()

df["clean_text"] = df["clean_text"].apply(clean_text)

print("\nOriginal text:")
print(df["transcription"].iloc[0][:500])

print("\nCleaned text:")
print(df["clean_text"].iloc[0][:500])

stop_words = "english"

vectorizer = TfidfVectorizer(
    stop_words=stop_words,
    max_features=5000,
    min_df=2,
    max_df=0.95
)

X_tfidf = vectorizer.fit_transform(
    df["clean_text"]
)

print("\nTF-IDF shape:", X_tfidf.shape)

feature_names = vectorizer.get_feature_names_out()

tfidf_sum = np.asarray(X_tfidf.sum(axis=0)).flatten()

word_frequency = pd.DataFrame({
    "word": feature_names,
    "score": tfidf_sum
}).sort_values(
    "score",
    ascending=False
)

print("\nTop words:")
display(word_frequency.head(20))

plt.figure(figsize=(10, 6))
sns.barplot(
    x="score",
    y="word",
    data=word_frequency.head(20)
)
plt.title("Top 20 TF-IDF Terms")
plt.xlabel("TF-IDF Score")
plt.ylabel("Word")
plt.show()

label_encoder = LabelEncoder()

y_text = label_encoder.fit_transform(
    df["medical_specialty"]
)

print("\nNumber of classes:", len(label_encoder.classes_))

X_train_text, X_test_text, y_train_text, y_test_text = train_test_split(
    X_tfidf,
    y_text,
    test_size=0.20,
    random_state=42,
    stratify=y_text
)

print("\nTraining text shape:", X_train_text.shape)
print("Testing text shape:", X_test_text.shape)

print("\nText preprocessing complete.")
```

**#Part-C: Image Data**

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from PIL import Image
from tensorflow.keras.preprocessing.image import ImageDataGenerator

import kagglehub

path = kagglehub.dataset_download(
    "paultimothymooney/chest-xray-pneumonia"
)

print("Path to dataset files:", path)

# The dataset files are often nested inside an extra 'chest_xray' directory.
dataset_path = path
if os.path.exists(os.path.join(path, "chest_xray")):
    dataset_path = os.path.join(path, "chest_xray")

records = []

for split in ["train", "test", "val"]:
    split_path = os.path.join(dataset_path, split)

    if os.path.exists(split_path):
        for label in ["NORMAL", "PNEUMONIA"]:
            label_path = os.path.join(split_path, label)

            if os.path.exists(label_path):
                for filename in os.listdir(label_path):
                    filepath = os.path.join(
                        label_path,
                        filename
                    )

                    if filename.lower().endswith(
                        (".jpg", ".jpeg", ".png")
                    ):
                        records.append({
                            "filepath": filepath,
                            "split": split,
                            "label": label
                        })

image_df = pd.DataFrame(records)

print("\nTotal images:", len(image_df))

display(image_df.head())

print("\nDataset splits:")
print(image_df["split"].value_counts())

print("\nClass distribution:")
print(image_df["label"].value_counts())

print("\nClass distribution by split:")
print(
    pd.crosstab(
        image_df["split"],
        image_df["label"]
    )
)

plt.figure(figsize=(7, 5))

sns.countplot(
    x="label",
    data=image_df
)

plt.title("Chest X-Ray Class Distribution")

plt.xlabel("Class")
plt.ylabel("Number of Images")

plt.show()

plt.figure(figsize=(8, 5))

sns.countplot(
    x="split",
    hue="label",
    data=image_df
)

plt.title("Class Distribution by Dataset Split")

plt.xlabel("Dataset Split")
plt.ylabel("Number of Images")

plt.show()

sample_images = image_df.sample(
    min(8, len(image_df)),
    random_state=42
)

plt.figure(figsize=(12, 8))

for i, (_, row) in enumerate(
    sample_images.iterrows()
):
    image = Image.open(row["filepath"])

    plt.subplot(2, 4, i + 1)
    plt.imshow(image, cmap="gray")
    plt.title(
        row["label"] + " (" + row["split"] + ")"
    )
    plt.axis("off")

plt.tight_layout()
plt.show()

image_sizes = []

sample_for_size = image_df.sample(
    min(500, len(image_df)),
    random_state=42
)

for filepath in sample_for_size["filepath"]:
    with Image.open(filepath) as image:
        image_sizes.append(image.size)

size_df = pd.DataFrame(
    image_sizes,
    columns=["width", "height"]
)

print("\nImage size statistics:")
display(size_df.describe())

plt.figure(figsize=(8, 5))

sns.scatterplot(
    x="width",
    y="height",
    data=size_df
)

plt.title("Image Dimensions")

plt.xlabel("Width")
plt.ylabel("Height")

plt.show()

plt.figure(figsize=(8, 5))

sns.histplot(
    size_df["width"],
    bins=30,
    kde=True
)

plt.title("Distribution of Image Widths")

plt.xlabel("Width")
plt.ylabel("Number of Images")

plt.show()

plt.figure(figsize=(8, 5))

sns.histplot(
    size_df["height"],
    bins=30,
    kde=True
)

plt.title("Distribution of Image Heights")

plt.xlabel("Height")
plt.ylabel("Number of Images")

plt.show()

image_size = (224, 224)

train_path = os.path.join(
    dataset_path,
    "train"
)

test_path = os.path.join(
    dataset_path,
    "test"
)

train_datagen = ImageDataGenerator(
    rescale=1.0 / 255.0,
    rotation_range=10,
    width_shift_range=0.05,
    height_shift_range=0.05,
    zoom_range=0.10,
    horizontal_flip=True
)

test_datagen = ImageDataGenerator(
    rescale=1.0 / 255.0
)

train_generator = train_datagen.flow_from_directory(
    train_path,
    target_size=image_size,
    color_mode="grayscale",
    class_mode="binary",
    batch_size=32,
    shuffle=True,
    seed=42
)

test_generator = test_datagen.flow_from_directory(
    test_path,
    target_size=image_size,
    color_mode="grayscale",
    class_mode="binary",
    batch_size=32,
    shuffle=False
)

X_image_batch, y_image_batch = next(
    train_generator
)

print("\nImage tensor shape:")
print(X_image_batch.shape)

print("\nImage label shape:")
print(y_image_batch.shape)

print("\nPixel minimum:")
print(X_image_batch.min())

print("\nPixel maximum:")
print(X_image_batch.max())

plt.figure(figsize=(12, 8))

for i in range(
    min(8, len(X_image_batch))
):
    plt.subplot(2, 4, i + 1)

    plt.imshow(
        X_image_batch[i].squeeze(),
        cmap="gray"
    )

    plt.title(
        "Pneumonia"
        if y_image_batch[i] == 1
        else "Normal"
    )

    plt.axis("off")

plt.tight_layout()
plt.show()

class_names = list(
    train_generator.class_indices.keys()
)

print("\nClass indices:")
print(train_generator.class_indices)

print("\nTraining images:",
      train_generator.samples)

print("Testing images:",
      test_generator.samples)

print("\nImage preprocessing complete.")
```

```python
!pip install wfdb -q

import wfdb
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.signal import butter, filtfilt

record_name = "100"

record = wfdb.rdrecord(record_name, pn_dir="mitdb")
annotation = wfdb.rdann(record_name, "atr", pn_dir="mitdb")

signal = record.p_signal[:, 0]
fs = record.fs

print("========== DATASET INFORMATION ==========")
print("Dataset: MIT-BIH Arrhythmia Database")
print("Record:", record_name)
print("Sampling frequency:", fs, "Hz")
print("Number of samples:", len(signal))
print("Number of channels:", record.n_sig)
print("Signal duration:", round(len(signal) / fs, 2), "seconds")

print("\n========== SIGNAL STATISTICS ==========")
print("Minimum:", round(np.min(signal), 4))
print("Maximum:", round(np.max(signal), 4))
print("Mean:", round(np.mean(signal), 4))
print("Standard deviation:", round(np.std(signal), 4))

time = np.arange(len(signal)) / fs
samples_to_plot = int(10 * fs)

plt.figure(figsize=(14, 5))
plt.plot(time[:samples_to_plot], signal[:samples_to_plot])
plt.title("Raw ECG Signal - MIT-BIH Record 100")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude (mV)")
plt.grid(True)
plt.show()

annotation_counts = pd.Series(annotation.symbol).value_counts()

print("\n========== ECG ANNOTATIONS ==========")
print("Total annotations:", len(annotation.symbol))
print("\nAnnotation distribution:")
print(annotation_counts)

plt.figure(figsize=(12, 5))
annotation_counts.head(15).plot(kind="bar")
plt.title("ECG Annotation Distribution")
plt.xlabel("Annotation Type")
plt.ylabel("Number of Annotations")
plt.xticks(rotation=45)
plt.show()

annotation_samples = annotation.sample[
    annotation.sample < samples_to_plot
]

annotation_values = signal[annotation_samples]

plt.figure(figsize=(14, 5))
plt.plot(
    time[:samples_to_plot],
    signal[:samples_to_plot],
    label="ECG"
)
plt.scatter(
    annotation_samples / fs,
    annotation_values,
    s=30,
    label="Beat annotations"
)
plt.title("ECG Signal with Beat Annotations")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude (mV)")
plt.legend()
plt.grid(True)
plt.show()

beat_times = annotation.sample / fs
rr_intervals = np.diff(beat_times)

rr_intervals_clean = rr_intervals[
    (rr_intervals > 0.25) &
    (rr_intervals < 2.0)
]

mean_rr = np.mean(rr_intervals_clean)
heart_rate = 60 / mean_rr

print("\n========== RR INTERVAL ANALYSIS ==========")
print("Total RR intervals:", len(rr_intervals))
print("Filtered RR intervals:", len(rr_intervals_clean))
print("Mean RR interval:", round(mean_rr, 3), "seconds")
print("Estimated heart rate:", round(heart_rate, 2), "BPM")

plt.figure(figsize=(12, 5))
plt.plot(rr_intervals_clean)
plt.title("Filtered RR Intervals")
plt.xlabel("Beat Number")
plt.ylabel("RR Interval (seconds)")
plt.grid(True)
plt.show()

def bandpass_filter(signal, lowcut, highcut, fs, order=4):
    nyquist = 0.5 * fs
    low = lowcut / nyquist
    high = highcut / nyquist
    b, a = butter(order, [low, high], btype="band")
    return filtfilt(b, a, signal)

filtered_signal = bandpass_filter(
    signal,
    0.5,
    40,
    fs
)

print("\n========== NOISE REMOVAL ==========")
print("Band-pass filter range: 0.5 - 40 Hz")
print("Noise filtering completed.")

plt.figure(figsize=(14, 6))
plt.plot(
    time[:samples_to_plot],
    signal[:samples_to_plot],
    label="Raw ECG"
)
plt.plot(
    time[:samples_to_plot],
    filtered_signal[:samples_to_plot],
    label="Filtered ECG"
)
plt.title("Raw vs Filtered ECG")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude (mV)")
plt.legend()
plt.grid(True)
plt.show()

window_size = int(2 * fs)
step_size = int(1 * fs)

windows = []
window_times = []

for start in range(
    0,
    len(filtered_signal) - window_size,
    step_size
):
    end = start + window_size
    windows.append(filtered_signal[start:end])
    window_times.append(start / fs)

X_signal = np.array(windows)

print("\n========== SIGNAL WINDOWING ==========")
print("Window duration: 2 seconds")
print("Window size:", window_size, "samples")
print("Step size:", step_size, "samples")
print("Number of windows:", X_signal.shape[0])
print("Samples per window:", X_signal.shape[1])
print("Signal data shape:", X_signal.shape)

plt.figure(figsize=(14, 8))

for i in range(4):
    plt.subplot(4, 1, i + 1)
    plt.plot(X_signal[i])
    plt.title("ECG Window " + str(i + 1))
    plt.xlabel("Sample")
    plt.ylabel("Amplitude")

plt.tight_layout()
plt.show()

window_mean = X_signal.mean(
    axis=1,
    keepdims=True
)

window_std = X_signal.std(
    axis=1,
    keepdims=True
)

X_signal_normalized = (
    X_signal - window_mean
) / (
    window_std + 1e-8
)

print("\n========== NORMALIZATION ==========")
print("Normalized signal shape:",
      X_signal_normalized.shape)

print("Mean of first window:",
      round(X_signal_normalized[0].mean(), 5))

print("Standard deviation of first window:",
      round(X_signal_normalized[0].std(), 5))

plt.figure(figsize=(14, 5))
plt.plot(
    X_signal[0],
    label="Original"
)
plt.plot(
    X_signal_normalized[0],
    label="Normalized"
)
plt.title("Original vs Normalized ECG Window")
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)
plt.show()

normal_symbols = {"N"}

abnormal_symbols = {
    "L", "R", "A", "a", "J", "S",
    "V", "F", "[", "]", "E", "j",
    "e", "/", "f", "Q"
}

window_labels = []

for start in range(
    0,
    len(filtered_signal) - window_size,
    step_size
):
    end = start + window_size

    nearby_annotations = []

    for i, sample in enumerate(annotation.sample):
        if start <= sample < end:
            nearby_annotations.append(
                annotation.symbol[i]
            )

    if len(nearby_annotations) == 0:
        window_labels.append(0)
    elif any(
        symbol in abnormal_symbols
        for symbol in nearby_annotations
    ):
        window_labels.append(1)
    else:
        window_labels.append(0)

y_signal = np.array(window_labels)

print("\n========== ECG LABELING ==========")
print("Class 0: Normal")
print("Class 1: Abnormal")
print("\nClass distribution:")
print(pd.Series(y_signal).value_counts())

plt.figure(figsize=(6, 4))
sns.countplot(x=y_signal)
plt.title("ECG Window Class Distribution")
plt.xlabel("Class (0 = Normal, 1 = Abnormal)")
plt.ylabel("Number of Windows")
plt.show()

noise = np.random.normal(
    loc=0,
    scale=0.02,
    size=X_signal_normalized.shape
)

X_signal_augmented = (
    X_signal_normalized + noise
)

print("\n========== DATA AUGMENTATION ==========")
print("Original shape:",
      X_signal_normalized.shape)
print("Augmented shape:",
      X_signal_augmented.shape)

plt.figure(figsize=(14, 5))
plt.plot(
    X_signal_normalized[0],
    label="Original"
)
plt.plot(
    X_signal_augmented[0],
    label="Augmented"
)
plt.title("ECG Signal Augmentation")
plt.xlabel("Sample")
plt.ylabel("Normalized Amplitude")
plt.legend()
plt.grid(True)
plt.show()

X_final = X_signal_normalized
y_final = y_signal

print("\n========== FINAL MODEL-READY DATA ==========")
print("Feature shape:", X_final.shape)
print("Target shape:", y_final.shape)
print("Missing values:", np.isnan(X_final).sum())
print("Infinite values:", np.isinf(X_final).sum())

print("\n========== FINAL SUMMARY ==========")
print("Data acquisition: Completed")
print("Data inspection: Completed")
print("EDA: Completed")
print("Annotation analysis: Completed")
print("RR interval analysis: Completed")
print("Heart-rate estimation: Completed")
print("Noise removal: Completed")
print("Signal filtering: Completed")
print("Signal windowing: Completed")
print("Normalization: Completed")
print("Class analysis: Completed")
print("Signal augmentation: Completed")
print("Model-ready preprocessing: Completed")
```
